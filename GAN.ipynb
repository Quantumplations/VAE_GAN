$\textbf{Problem}$: Explain why $\textit{generative adversarial networks}$ (GANs) represent a unique kind of $\textit{training}$ paradigm compared to conventional supervised cost function gradient descent.

$\textbf{Solution}$: At a purely theoretical level, the cost function to be $\textit{minimaxed}$ (not just minimized!) is:
$$C(\boldsymbol{\theta},\boldsymbol{\phi})=\langle\ln D_{\boldsymbol{\phi}}(\mathbf x)\rangle+\langle\ln (1-D_{\boldsymbol{\phi}}(G_{\boldsymbol{\theta}}(\boldsymbol{\varepsilon})))\rangle$$

where:

$$\langle\ln D_{\boldsymbol{\phi}}(\mathbf x)\rangle:=\int d\mathbf xp(\mathbf x)\ln D_{\boldsymbol{\phi}}(\mathbf x)$$

$$\langle\ln (1-D_{\boldsymbol{\phi}}(G_{\boldsymbol{\theta}}(\boldsymbol{\varepsilon})))\rangle:=\int d\boldsymbol{\varepsilon}p(\boldsymbol{\varepsilon})\ln (1-D_{\boldsymbol{\phi}}(G_{\boldsymbol{\theta}}(\boldsymbol{\varepsilon})))$$

$\textbf{Problem}$: Above, it was explained how GANs are used during training. What about at inference time?

$\textbf{Solution}$: At inference time, throw away the discriminator! Just use the generator, sampling a random latent $\boldsymbol{\varepsilon}$ and feeding into the trained generator $\boldsymbol{G}_{\boldsymbol{\theta}}(\boldsymbol{\varepsilon})$ to sample from the data distribution $p(\mathbf x)$ (this is analogous throwing away the encoder $e_{\boldsymbol{\phi}}$ and just using the decoder $d_{\boldsymbol{\theta}}$ for inference-time generation in autoencoders).
